```python
import pandas as pd
import numpy as np

file_1 = directory + "/1year.arff"
file_1_data = pd.read_csv(file_1, header=None)
```

```python
from sklearn.datasets import load_digits
data = load_digits()
```

```python
x = data['data']
y = data['target']
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(n_estimators=20)
score = cross_val_score(rf_model, x, y, cv=5, scoring='accuracy')
```

```python
# using the random search on the digits dataset to get the numbers tuned up

from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(n_estimators=20, random_state=42)
param_dist = {'max_depth': [3, None],
              'max_features': randint(1, 11),
              'min_samples_split': randint(2, 11),
              'min_samples_leaf': randint(1, 11),
              'criterion': ['gini', 'entropy']}

rf_random = RandomizedSearchCV(estimator=rf_model, param_distributions=param_dist,  n_iter=100, cv=5, verbose=2, random_state=42, n_jobs=-1)
rf_random.fit(x, y)



```
